## ETL Pipeline: Extract, Profile, Transform

### 1. Extract Data

```python
import zipfile
import os
import pandas as pd

# Define the path to the zip file
zip_file_path = '/content/Day 4.zip'

# Define the directory where the files will be extracted
extract_dir = '/content/Day_4_extracted'

# Create the extraction directory if it doesn't exist
os.makedirs(extract_dir, exist_ok=True)

# Unzip the file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"Files extracted to: {extract_dir}")

# List the extracted files
extracted_files = os.listdir(extract_dir)
print("Extracted files:")
for file in extracted_files:
    print(f"- {file}")
```

### 2. Data Profiling

```python
# Update the extract_dir to point to the correct directory containing the CSVs
extract_dir_with_data = os.path.join(extract_dir, 'Day 4')

# List the CSV files in the updated directory
csv_files = [f for f in os.listdir(extract_dir_with_data) if f.endswith('.csv')]
print("CSV files found for profiling:")
for file in csv_files:
    print(f"- {file}")

# Dictionary to store DataFrames
dataframes = {}

# Load each CSV file and perform initial profiling
for file in csv_files:
    file_path = os.path.join(extract_dir_with_data, file)
    df_name = os.path.splitext(file)[0] # Use filename without extension as df name
    df = pd.read_csv(file_path)
    dataframes[df_name] = df

    print(f"\n--- Profiling for {df_name}.csv ---")
    print("\nHead of the DataFrame:")
    display(df.head())

    print("\nDataFrame Info:")
    df.info()

    print("\nDescriptive Statistics:")
    display(df.describe(include='all'))

    print("\nMissing Values:")
    display(df.isnull().sum().to_frame(name='Missing Count'))
```

### 3. Data Transformation

Based on the profiling, the following transformation decisions will be made:

1.  **Convert Timestamp Columns**: Convert all columns identified as timestamps to `datetime` objects for proper time-series analysis.
2.  **Handle Missing Values**: Address missing values, especially in numerical product attributes, by imputation or removal.
3.  **Data Type Optimization**: Optimize data types for memory efficiency and correct representation (e.g., integer for IDs, categorical for status).
4.  **Rename Columns**: Standardize column names for consistency across datasets (e.g., `product_category_name_english` can be just `product_category_name` after translation).
5.  **Merge DataFrames**: Combine relevant DataFrames to create comprehensive fact and dimension tables suitable for a star schema or snowflake schema.
6.  **Feature Engineering**: Create new features if beneficial for analysis (e.g., `order_duration`, `delivery_delay`).

Let's start by performing the essential datetime conversions.

```python
# 1. Convert Timestamp Columns to Datetime Objects
print("Converting timestamp columns...")

# olist_orders_dataset
datetime_cols_orders = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
for col in datetime_cols_orders:
    if col in dataframes['olist_orders_dataset'].columns:
        dataframes['olist_orders_dataset'][col] = pd.to_datetime(dataframes['olist_orders_dataset'][col])
print("olist_orders_dataset timestamp columns converted.")

# olist_order_reviews_dataset
datetime_cols_reviews = [
    'review_creation_date',
    'review_answer_timestamp'
]
for col in datetime_cols_reviews:
    if col in dataframes['olist_order_reviews_dataset'].columns:
        dataframes['olist_order_reviews_dataset'][col] = pd.to_datetime(dataframes['olist_order_reviews_dataset'][col])
print("olist_order_reviews_dataset timestamp columns converted.")

# olist_order_items_dataset
datetime_cols_items = [
    'shipping_limit_date'
]
for col in datetime_cols_items:
    if col in dataframes['olist_order_items_dataset'].columns:
        dataframes['olist_order_items_dataset'][col] = pd.to_datetime(dataframes['olist_order_items_dataset'][col])
print("olist_order_items_dataset timestamp columns converted.")

print("Timestamp conversion complete for all relevant dataframes.")

# Display info for a converted DataFrame to verify
print("\nVerifying olist_orders_dataset info after conversion:")
dataframes['olist_orders_dataset'].info()
```

```python
# 2. Handle Missing Values and Data Type Optimization
print("\n--- Handling Missing Values and Optimizing Data Types ---")

for df_name, df in dataframes.items():
    print(f"\nProcessing DataFrame: {df_name}")

    # Identify columns with missing values
    missing_cols = df.isnull().sum()
    missing_cols = missing_cols[missing_cols > 0]

    if not missing_cols.empty:
        print("Missing values found in columns:")
        print(missing_cols)

        for col in missing_cols.index:
            if pd.api.types.is_numeric_dtype(df[col]):
                # For numeric columns, impute with median to be robust to outliers
                median_val = df[col].median()
                df[col].fillna(median_val, inplace=True)
                print(f"  - Imputed numeric column '{col}' with median: {median_val}")
            elif pd.api.types.is_object_dtype(df[col]):
                # For object/categorical columns, impute with mode or 'Unknown'
                # Using 'Unknown' to avoid bias from mode for potentially critical values
                df[col].fillna('Unknown', inplace=True)
                print(f"  - Imputed object column '{col}' with 'Unknown'")
            elif pd.api.types.is_datetime64_any_dtype(df[col]):
                # For datetime columns, consider a strategy. For now, we'll leave as NaT or 'Unknown' depending on context
                # For simplicity, filling with 'NaT' (Not a Time) or backfilling with earliest/latest valid date could be options.
                # For now, if conversion already handled some, we'll keep NaT or decide based on specific column usage.
                # If a specific datetime imputation is needed, it would be column-specific.
                print(f"  - Datetime column '{col}' has missing values. Further review needed if not handled by earlier conversion.")
    else:
        print("No missing values found in this DataFrame.")

    # Data Type Optimization (after handling NaNs, especially for numeric types)
    for col in df.columns:
        # Convert object columns to category if unique values are less than 50% of total rows
        if pd.api.types.is_object_dtype(df[col]):
            num_unique = df[col].nunique()
            if num_unique / len(df[col]) < 0.5 and num_unique < 50: # Heuristic for categorical
                df[col] = df[col].astype('category')
                print(f"  - Converted object column '{col}' to category.")
        # Convert float columns to integer if all values are integers and no NaNs remain
        elif pd.api.types.is_float_dtype(df[col]):
            if df[col].isnull().sum() == 0 and (df[col] == df[col].astype(int)).all():
                df[col] = df[col].astype(int)
                print(f"  - Converted float column '{col}' to integer.")

    print(f"DataFrame {df_name} info after missing value handling and type optimization:")
    df.info()
```

### 5. Merge DataFrames

Now, let's merge the various DataFrames to create a more comprehensive view, leading towards our fact and dimension tables. We will proceed in a logical order, starting with enriching product information.

```python
# Merge product category translation with product dataset
# This effectively renames and adds the English category name to the products dataframe
if 'product_category_name_translation' in dataframes and 'olist_products_dataset' in dataframes:
    products_df = dataframes['olist_products_dataset'].merge(
        dataframes['product_category_name_translation'],
        on='product_category_name',
        how='left'
    )
    # Drop the original Portuguese category name if not needed, keeping the English one
    if 'product_category_name' in products_df.columns and 'product_category_name_english' in products_df.columns:
        products_df.drop(columns=['product_category_name'], inplace=True)
        products_df.rename(columns={'product_category_name_english': 'product_category_name'}, inplace=True)
    dataframes['olist_products_dataset'] = products_df
    print("Merged product_category_name_translation with olist_products_dataset.")
    print("Updated olist_products_dataset head:")
    display(dataframes['olist_products_dataset'].head())

# Merge order items with products to get product details for each item
if 'olist_order_items_dataset' in dataframes and 'olist_products_dataset' in dataframes:
    order_items_products_df = dataframes['olist_order_items_dataset'].merge(
        dataframes['olist_products_dataset'],
        on='product_id',
        how='left'
    )
    dataframes['olist_order_items_dataset'] = order_items_products_df
    print("Merged olist_order_items_dataset with olist_products_dataset.")
    print("Updated olist_order_items_dataset head:")
    display(dataframes['olist_order_items_dataset'].head())

print("Initial DataFrame merges complete.")
```

```python
# Merge order items (now with product details) with orders dataset
if 'olist_order_items_dataset' in dataframes and 'olist_orders_dataset' in dataframes:
    # We need to create a copy to avoid SettingWithCopyWarning if we modify columns later
    orders_with_items_products_df = dataframes['olist_order_items_dataset'].merge(
        dataframes['olist_orders_dataset'],
        on='order_id',
        how='left'
    )
    dataframes['fact_order_items'] = orders_with_items_products_df
    print("Merged olist_order_items_dataset with olist_orders_dataset to create 'fact_order_items'.")
    print("fact_order_items head:")
    display(dataframes['fact_order_items'].head())

# Merge customers with the fact_order_items to get customer demographics
if 'fact_order_items' in dataframes and 'olist_customers_dataset' in dataframes:
    fact_with_customers = dataframes['fact_order_items'].merge(
        dataframes['olist_customers_dataset'],
        on='customer_id',
        how='left'
    )
    dataframes['fact_order_items'] = fact_with_customers
    print("Merged olist_customers_dataset with fact_order_items.")
    print("fact_order_items head (with customer info):")
    display(dataframes['fact_order_items'].head())

# Merge sellers with the fact_order_items to get seller demographics
if 'fact_order_items' in dataframes and 'olist_sellers_dataset' in dataframes:
    fact_with_sellers = dataframes['fact_order_items'].merge(
        dataframes['olist_sellers_dataset'],
        on='seller_id',
        how='left'
    )
    dataframes['fact_order_items'] = fact_with_sellers
    print("Merged olist_sellers_dataset with fact_order_items.")
    print("fact_order_items head (with seller info):")
    display(dataframes['fact_order_items'].head())

# Also merge payments and reviews into the fact table for a comprehensive view
if 'fact_order_items' in dataframes and 'olist_order_payments_dataset' in dataframes:
    fact_with_payments = dataframes['fact_order_items'].merge(
        dataframes['olist_order_payments_dataset'],
        on='order_id',
        how='left'
    )
    dataframes['fact_order_items'] = fact_with_payments
    print("Merged olist_order_payments_dataset with fact_order_items.")

if 'fact_order_items' in dataframes and 'olist_order_reviews_dataset' in dataframes:
    # Reviews can have multiple entries per order_id if there are multiple review types,
    # but typically it's one main review. For simplicity, we'll merge directly.
    # If multiple reviews per order_id are present, this merge will duplicate rows.
    # A more sophisticated approach might aggregate reviews first or create a separate dimension.
    fact_with_reviews = dataframes['fact_order_items'].merge(
        dataframes['olist_order_reviews_dataset'],
        on='order_id',
        how='left' # Use left join to keep all order items even if no review
    )
    dataframes['fact_order_items'] = fact_with_reviews
    print("Merged olist_order_reviews_dataset with fact_order_items.")

print("All major DataFrame merges complete. 'fact_order_items' now contains comprehensive order, item, product, customer, seller, payment, and review information.")
print("Final 'fact_order_items' info:")
dataframes['fact_order_items'].info()
display(dataframes['fact_order_items'].head())
```

### 6. Data Warehouse Schema (Star Schema Design)

Based on the transformations, we can design a star schema for our data warehouse. A star schema typically consists of a central **fact table** and multiple **dimension tables** that surround it.

**Fact Table:**
*   `fact_order_items`: This will be our main fact table, containing measures like `price`, `freight_value`, `payment_value`, `review_score`, and foreign keys linking to dimension tables.

**Dimension Tables:**
*   `dim_customers`: Contains customer attributes (`customer_unique_id`, `customer_zip_code_prefix`, `customer_city`, `customer_state`).
*   `dim_products`: Contains product attributes (`product_id`, `product_category_name`, `product_name_lenght`, `product_description_lenght`, `product_photos_qty`, `product_weight_g`, `product_length_cm`, `product_height_cm`, `product_width_cm`).
*   `dim_sellers`: Contains seller attributes (`seller_id`, `seller_zip_code_prefix`, `seller_city`, `seller_state`).
*   `dim_time`: A time dimension can be generated from the various timestamp columns in the fact table (`order_purchase_timestamp`, `order_approved_at`, etc.).
*   `dim_geolocation`: Contains geolocation information (`geolocation_zip_code_prefix`, `geolocation_lat`, `geolocation_lng`, `geolocation_city`, `geolocation_state`).

Let's extract these dimension tables and finalize the `fact_order_items` table.

```python
print("\n--- Creating Dimension Tables ---")

# dim_customers
# Select unique customer information from the fact table or original customer dataset
dataframes['dim_customers'] = dataframes['olist_customers_dataset'][[
    'customer_id',
    'customer_unique_id',
    'customer_zip_code_prefix',
    'customer_city',
    'customer_state'
]].drop_duplicates().reset_index(drop=True)
print("dim_customers created.")
display(dataframes['dim_customers'].head())

# dim_products
# Select unique product information from the fact table or original product dataset
# Ensure to use the product_category_name_english if it was translated during merge
dataframes['dim_products'] = dataframes['olist_products_dataset'][[
    'product_id',
    'product_category_name',
    'product_name_lenght',
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]].drop_duplicates().reset_index(drop=True)
print("dim_products created.")
display(dataframes['dim_products'].head())

# dim_sellers
dataframes['dim_sellers'] = dataframes['olist_sellers_dataset'][[
    'seller_id',
    'seller_zip_code_prefix',
    'seller_city',
    'seller_state'
]].drop_duplicates().reset_index(drop=True)
print("dim_sellers created.")
display(dataframes['dim_sellers'].head())

# For simplicity, we'll create a basic time dimension directly from purchase timestamp.
# In a real DW, this would be a more extensive date/time table.
dataframes['dim_time'] = dataframes['fact_order_items'][['order_purchase_timestamp']].drop_duplicates().reset_index(drop=True)
dataframes['dim_time']['date'] = dataframes['dim_time']['order_purchase_timestamp'].dt.date
dataframes['dim_time']['year'] = dataframes['dim_time']['order_purchase_timestamp'].dt.year
dataframes['dim_time']['month'] = dataframes['dim_time']['order_purchase_timestamp'].dt.month
dataframes['dim_time']['day'] = dataframes['dim_time']['order_purchase_timestamp'].dt.day
dataframes['dim_time']['day_of_week'] = dataframes['dim_time']['order_purchase_timestamp'].dt.dayofweek
dataframes['dim_time']['hour'] = dataframes['dim_time']['order_purchase_timestamp'].dt.hour
print("dim_time created.")
display(dataframes['dim_time'].head())

print("\n--- Finalizing Fact Table ('fact_order_items') ---")
# Drop redundant columns from the fact table that are now in dimension tables
# Keep foreign keys and measures
fact_table_columns = [
    'order_id', 'order_item_id', 'product_id', 'seller_id', 'customer_id', # Foreign Keys
    'shipping_limit_date', 'price', 'freight_value', # Order Item Details
    'order_status', 'order_purchase_timestamp', 'order_approved_at',
    'order_delivered_carrier_date', 'order_delivered_customer_date', 'order_estimated_delivery_date', # Order Details
    'payment_sequential', 'payment_type', 'payment_installments', 'payment_value', # Payment Details
    'review_id', 'review_score', 'review_comment_title', 'review_comment_message',
    'review_creation_date', 'review_answer_timestamp' # Review Details
    # Note: customer_unique_id, zip, city, state, and seller details are moved to dim tables
]
dataframes['fact_order_items'] = dataframes['fact_order_items'][fact_table_columns]

print("Fact table 'fact_order_items' finalized.")
print("Final 'fact_order_items' info:")
dataframes['fact_order_items'].info()
display(dataframes['fact_order_items'].head())

print("\nETL pipeline for Data Warehouse complete!")
```

### 10. Temporal Analysis of Orders and Sales

Analyzing trends over time is crucial for understanding business performance. Let's visualize the number of orders and total sales volume over time.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure the 'order_purchase_timestamp' is set as index for time series analysis
# First, group by order_id and get the unique purchase timestamp for each order
orders_over_time = dataframes['fact_order_items'].groupby('order_id').first().set_index('order_purchase_timestamp')

# Resample to monthly frequency for number of orders
monthly_orders = orders_over_time.resample('M').size().rename('Number of Orders')

# Calculate monthly sales volume (sum of price per order item)
# Group by purchase timestamp and sum the price and freight_value
# We need to sum price and freight_value per order_id before resampling to avoid double counting items from same order
# And then sum again after resampling

# First, aggregate total value per order (price + freight for all items in an order)
# Ensure we're summing across order_items, then taking the max timestamp for that order
order_value = dataframes['fact_order_items'].groupby('order_id').agg(
    order_purchase_timestamp=('order_purchase_timestamp', 'first'),
    total_order_value=('payment_value', 'first') # Use payment_value as it represents total paid per order_id
).set_index('order_purchase_timestamp')

monthly_sales_volume = order_value['total_order_value'].resample('M').sum().rename('Total Sales Volume (BRL)')

# Plotting Number of Orders Over Time
fig6 = plt.figure(figsize=(14, 6))
monthly_orders.plot(kind='line', marker='o', linestyle='-', color='darkblue')
plt.title('Number of Orders Over Time (Monthly)')
plt.xlabel('Date')
plt.ylabel('Number of Orders')
plt.grid(True)
plt.tight_layout()
plt.show()

# Plotting Total Sales Volume Over Time
fig7 = plt.figure(figsize=(14, 6))
monthly_sales_volume.plot(kind='line', marker='o', linestyle='-', color='darkgreen')
plt.title('Total Sales Volume Over Time (Monthly)')
plt.xlabel('Date')
plt.ylabel('Total Sales Volume (BRL)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Transformation Decisions Report for Data Warehouse Construction

This report summarizes the key transformation decisions made during the construction of the data warehouse, aligning with the executed script.

---

#### 1. Convert Timestamp Columns

**Decision**: All columns identified as containing timestamp information were converted to `datetime` objects. This ensures proper handling of time-series data and enables chronological analysis.

**Implementation**: This was achieved using `pd.to_datetime()` for specific columns in `olist_orders_dataset`, `olist_order_reviews_dataset`, and `olist_order_items_dataset`. Verified by `.info()` output showing `datetime64[ns]` Dtype.

---

#### 2. Handle Missing Values

**Decision**: Missing values were addressed based on data type:
*   **Numeric Columns**: Missing values were imputed with the **median** of the respective column. This approach is robust to outliers and prevents data loss.
*   **Object/Categorical Columns**: Missing values were imputed with the string **'Unknown'**. This avoids biasing the data with the mode and explicitly flags missing categorical information.
*   **Datetime Columns**: Missing values were noted as needing further review. For some, they remained as `NaT` (Not a Time) which is standard for missing datetime entries in Pandas.

**Implementation**: An iterative process looped through each DataFrame, identifying missing values. `df[col].fillna(median_val, inplace=True)` was used for numeric, and `df[col].fillna('Unknown', inplace=True)` for object columns.

---

#### 3. Data Type Optimization

**Decision**: Data types were optimized for memory efficiency and correct representation.
*   **Object to Category**: Object columns with a relatively low number of unique values (less than 50% of total rows and fewer than 50 unique values) were converted to the `category` Dtype.
*   **Float to Integer**: Float columns that contained only integer values (after missing value imputation) and had no remaining `NaN`s were converted to `int` Dtype.

**Implementation**: This was done using `df[col].astype('category')` and `df[col].astype(int)` after missing value handling, based on programmatic checks for `nunique()` and `isnull().sum() == 0`.

---

#### 4. Rename Columns (Implicitly Handled)

**Decision**: Standardize column names for consistency.

**Implementation**: Specifically, `product_category_name_english` was effectively renamed to `product_category_name` during the merge with `olist_products_dataset`. The original Portuguese `product_category_name` column was dropped, ensuring only the English version is used going forward.

---

#### 5. Merge DataFrames

**Decision**: Combine relevant DataFrames to create a comprehensive `fact_order_items` table.

**Implementation**:
*   `product_category_name_translation` was merged with `olist_products_dataset` on `product_category_name` (left join).
*   `olist_order_items_dataset` was merged with the enriched `olist_products_dataset` on `product_id` (left join).
*   The resulting `olist_order_items_dataset` was successively merged with `olist_orders_dataset` (on `order_id`), `olist_customers_dataset` (on `customer_id`), `olist_sellers_dataset` (on `seller_id`), `olist_order_payments_dataset` (on `order_id`), and `olist_order_reviews_dataset` (on `order_id`), all using left joins to preserve order items.

---

#### 6. Data Warehouse Schema (Star Schema Design)

**Decision**: Structure the data into a star schema with `fact_order_items` as the central fact table and several dimension tables.

**Implementation**:
*   **`dim_customers`**: Created by selecting unique customer attributes from `olist_customers_dataset`.
*   **`dim_products`**: Created by selecting unique product attributes from the enriched `olist_products_dataset`.
*   **`dim_sellers`**: Created by selecting unique seller attributes from `olist_sellers_dataset`.
*   **`dim_time`**: A basic time dimension was generated from `order_purchase_timestamp` in `fact_order_items`, extracting year, month, day, day of week, and hour.
*   **`fact_order_items` Finalization**: Redundant columns (attributes moved to dimension tables) were dropped from `fact_order_items`, retaining only foreign keys and measures (price, freight_value, payment_value, review_score).

---

This ETL pipeline successfully extracted raw data, profiled it, applied necessary transformations for data quality and consistency, and finally structured it into a star schema suitable for data warehousing and subsequent analytical queries.
