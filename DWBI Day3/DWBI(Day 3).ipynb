```python
import pandas as pd
import os
from ydata_profiling import ProfileReport

# ============================================
# 1. LOAD DATASET
# ============================================

file_path = "/content/2020_al_data_kaggle_upload_new_old_syllabi.csv"

if os.path.exists(file_path):

    df = pd.read_csv(file_path)

    print(f"Dataset loaded successfully from:")
    print(file_path)

    # ============================================
    # 2. BASIC DATASET INFORMATION
    # ============================================

    print("=" * 60)
    print("DATASET PROFILING")
    print("=" * 60)

    print("\nDataset Shape:")
    print("Rows    :", df.shape[0])
    print("Columns :", df.shape[1])

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    # ============================================
    # 3. MISSING VALUES
    # ============================================

    print("\n" + "=" * 60)
    print("MISSING VALUES")
    print("=" * 60)

    missing = df.isnull().sum()

    missing_percentage = (missing / len(df)) * 100

    missing_report = pd.DataFrame({
        "Missing Values": missing,
        "Percentage": missing_percentage.round(2)
    })

    print(missing_report)

    # ============================================
    # 4. DUPLICATE RECORDS
    # ============================================

    print("\n" + "=" * 60)
    print("DUPLICATE RECORDS")
    print("=" * 60)

    duplicate_count = df.duplicated().sum()

    print("Number of duplicate records:", duplicate_count)

    # ============================================
    # 5. UNIQUE VALUES
    # ============================================

    print("\n" + "=" * 60)
    print("UNIQUE VALUES")
    print("=" * 60)

    print(df.nunique())

    # ============================================
    # 6. NUMERICAL STATISTICS
    # ============================================

    print("\n" + "=" * 60)
    print("NUMERICAL STATISTICS")
    print("=" * 60)

    print(df.describe())

    # ============================================
    # 7. CATEGORICAL VARIABLES
    # ============================================

    print("\n" + "=" * 60)
    print("CATEGORICAL VARIABLES")
    print("=" * 60)

    categorical_columns = df.select_dtypes(
        include=["object", "category"]
    ).columns

    for column in categorical_columns:

        print("\nColumn:", column)

        print(df[column].value_counts(dropna=False))

    # ============================================
    # 8. AUTOMATIC PROFILING REPORT
    # ============================================

    print("\nGenerating profiling report...")

    profile = ProfileReport(
        df,
        title="2020 A/L Dataset Profiling Report",
        explorative=True
    )

    profile.to_file("/content/dataset_profile.html")

    # ============================================
    # 9. COMPLETION MESSAGE
    # ============================================

    print("\n" + "=" * 60)
    print("PROFILING COMPLETED SUCCESSFULLY!")
    print("=" * 60)

    print("Report saved as:")
    print("/content/dataset_profile.html")

else:

    print("ERROR: Dataset not found!")

    print("\nPlease upload:")
    print("2020_al_data_kaggle_upload_new_old_syllabi.csv")

    print("\nIn Google Colab:")
    print("1. Click the folder icon on the left.")
    print("2. Click the Upload button.")
    print("3. Select your CSV file.")
    print("4. Run this cell again.")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = "/content/filtered_students_data.csv"

df = pd.read_csv(file_path)

print("=" * 60)
print("ORIGINAL DATASET")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 2. REMOVE UNNECESSARY INDEX COLUMN
# ============================================================

if "df_index" in df.columns:
    df = df.drop(columns=["df_index"])

print("\nRemoved df_index column.")


# ============================================================
# 3. REMOVE DUPLICATE RECORDS
# ============================================================

before = len(df)

df = df.drop_duplicates()

after = len(df)

print(f"Duplicates removed: {before - after}")


# ============================================================
# 4. CLEAN STRING COLUMNS
# ============================================================

string_columns = df.select_dtypes(include="object").columns

for col in string_columns:
    df[col] = df[col].astype(str).str.strip()

    # Convert common missing-value representations to NaN
    df[col] = df[col].replace(
        ["", "NA", "N/A", "na", "n/a", "NULL", "null", "-", "--"],
        np.nan
    )


# ============================================================
# 5. CONVERT NUMERIC COLUMNS
# ============================================================

numeric_columns = [
    "Zscore",
    "district_rank",
    "island_rank",
    "al_year",
    "sub1_r",
    "sub2_r",
    "sub3_r",
    "cgt_r",
    "ge_r",
    "birth_day",
    "birth_year"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")


# ============================================================
# 6. CONVERT BIRTH MONTH TO NUMBER
# ============================================================

month_mapping = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

if "birth_month" in df.columns:
    df["birth_month_num"] = (
        df["birth_month"]
        .map(month_mapping)
    )


# ============================================================
# 7. CONVERT SUBJECT GRADES TO NUMERIC VALUES
# ============================================================

# A/L grade hierarchy
grade_mapping = {
    "A": 5,
    "B": 4,
    "C": 3,
    "S": 2,
    "W": 0,
    "F": 0
}

grade_columns = ["sub1", "sub2", "sub3"]

for col in grade_columns:
    if col in df.columns:
        df[col + "_encoded"] = (
            df[col]
            .str.upper()
            .map(grade_mapping)
        )


# ============================================================
# 8. ENCODE GENDER
# ============================================================

gender_mapping = {
    "male": 0,
    "female": 1
}

if "gender" in df.columns:
    df["gender_encoded"] = (
        df["gender"]
        .str.lower()
        .map(gender_mapping)
    )


# ============================================================
# 9. ENCODE STREAM
# ============================================================

if "stream" in df.columns:

    stream_encoder = LabelEncoder()

    df["stream_encoded"] = stream_encoder.fit_transform(
        df["stream"].astype(str)
    )

    print("\nStream Encoding:")
    for value, encoded in zip(
        stream_encoder.classes_,
        range(len(stream_encoder.classes_))
    ):
        print(value, "=", encoded)


# ============================================================
# 10. ENCODE SYLLABUS
# ============================================================

if "syllabus" in df.columns:

    syllabus_encoder = LabelEncoder()

    df["syllabus_encoded"] = syllabus_encoder.fit_transform(
        df["syllabus"].astype(str)
    )

    print("\nSyllabus Encoding:")
    for value, encoded in zip(
        syllabus_encoder.classes_,
        range(len(syllabus_encoder.classes_))
    ):
        print(value, "=", encoded)


# ============================================================
# 11. CREATE AGE AT A/L YEAR
# ============================================================

if "al_year" in df.columns and "birth_year" in df.columns:

    df["age_at_al"] = (
        df["al_year"] - df["birth_year"]
    )

    # Remove impossible ages
    df.loc[
        (df["age_at_al"] < 15) |
        (df["age_at_al"] > 30),
        "age_at_al"
    ] = np.nan


# ============================================================
# 12. HANDLE MISSING NUMERIC VALUES
# ============================================================

numeric_cols = df.select_dtypes(
    include=["int64", "float64"]
).columns

for col in numeric_cols:

    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(
            df[col].median()
        )


# ============================================================
# 13. HANDLE MISSING CATEGORICAL VALUES
# ============================================================

categorical_cols = df.select_dtypes(
    include=["object"]
).columns

for col in categorical_cols:

    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna("Unknown")


# ============================================================
# 14. CHECK TRANSFORMED DATA
# ============================================================

print("\n" + "=" * 60)
print("TRANSFORMED DATASET")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# 15. SAVE TRANSFORMED DATASET
# ============================================================

output_file = "/content/transformed_students_data.csv"

df.to_csv(
    output_file,
    index=False
)

print("\n" + "=" * 60)
print("TRANSFORMATION COMPLETED")
print("=" * 60)

print("Saved to:")
print(output_file)


# ============================================================
# 16. DISPLAY FIRST 10 RECORDS
# ============================================================

print("\nFirst 10 transformed records:")

display(df.head(10))
```
